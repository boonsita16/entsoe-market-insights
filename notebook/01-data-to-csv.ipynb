```python
from pathlib import Path
import pandas as pd
from entsoe import EntsoePandasClient
from entsoe.mappings import Area, lookup_area

# 1. Force fetch data for the demo period
cwd = Path.cwd()
api_file = cwd.parent.parent / "entsoe-api.txt"
if not api_file.exists():
    api_file = Path("entsoe-api.txt")

API_TOKEN = api_file.read_text(encoding="utf-8").strip()
client = EntsoePandasClient(api_key=API_TOKEN)

# Configuration for Demo generation
country_code = 'DK_1'
country_code_from = 'DK_1'
country_code_to = 'DE_LU'
local_tz = lookup_area(country_code).tz

start = pd.Timestamp('20260920', tz=local_tz)
end = pd.Timestamp('20260925', tz=local_tz)

print("⏳ Fetching demo data from ENTSO-E...")
try:
    df_price = client.query_day_ahead_prices(country_code, start=start, end=end)
    df_actual = client.query_generation(country_code, start=start, end=end, psr_type=None)
    df_load = client.query_load(country_code, start=start, end=end)
    df_net_pos = client.query_net_position(country_code, start=start, end=end)
    df_flows = client.query_crossborder_flows(country_code_from, country_code_to, start=start, end=end)

    # Save to CSV files
    df_price.to_csv("demo_day_ahead_prices.csv")
    df_actual.to_csv("demo_generation.csv")
    df_load.to_csv("demo_load.csv")
    df_net_pos.to_csv("demo_net_position.csv")
    df_flows.to_csv("demo_crossborder_flows.csv")
    print("✅ Demo CSV files saved successfully! You can now commit these to GitHub.")
except Exception as e:
    print(f"⚠️ Error preparing demo data: {e}")
```
